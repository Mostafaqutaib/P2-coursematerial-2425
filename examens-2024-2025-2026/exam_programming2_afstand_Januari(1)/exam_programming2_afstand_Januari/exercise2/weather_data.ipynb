## **Weather Data Analysis**

You are tasked with analyzing weather data collected over several days. Use the provided dataset with the following columns:
- `DATE`: The date of weather observation.
- `CITY`: The name of the city where the data was recorded.
- `TEMPERATURE`: The recorded temperature (in Celsius) for that day.

### **Tasks:**

1. **Import the necessary libraries.**

2. **Data Preprocessing:**
   - Load the weather dataset into a Pandas DataFrame.
   - Display the first 5 rows of the dataset.
   - Provide summary statistics for numeric columns (`TEMPERATURE`), including count, mean, standard deviation, minimum, and quartiles.

3. **Date Conversion:**
   - Convert the `DATE` column to pandas datetime objects.
   - Filter the data to include only rows from the year **2020**.

4. **Handle Missing Data:**
   - Replace any `NaN` values in the DataFrame with the mean temperature of the respective city.

5. **Visualization:**
   - Create a line plot showing the temperature trends over time.
   - Plot the `DATE` on the x-axis and `TEMPERATURE` on the y-axis.
   - Customize the plot with the following:
     - Use **red** as the color of the line.
     - Add a title (`Temperature Trends in 2020`) and appropriate axis labels (`Date` and `Temperature (°C)`).
     - Show a grid for better readability.
     - Include a legend indicating it represents "Temperature".
