## **Sales Data Analysis**

You are tasked with analyzing sales data collected across several stores for different products. Use the provided dataset with the following columns:
- `DATE`: The date of the sale.
- `STORE`: The name of the store where the sale took place.
- `PRODUCT`: The name of the product sold.
- `SALES`: The sales revenue (in USD) generated on that day.



### **Tasks:**

1. **Import the necessary libraries.**



2. **Data Preprocessing:**
   - Load the sales dataset into a Pandas DataFrame.
   - Display the first 5 rows of the dataset.
   - Provide summary statistics for numeric columns (`SALES`), including count, mean, standard deviation, minimum, and quartiles.


3. **Date Conversion:**
   - Convert the `DATE` column to pandas datetime objects.
   - Filter the data to include only rows from the year **2021**.


4. **Handle Missing Data:**
   - Replace any `NaN` values in the `SALES` column with the average sales revenue for the respective store.


5. **Visualization:**
   - Create a bar plot to compare the total sales revenue for each store.
   - Customize the plot with the following:
     - Use **blue** as the color of the bars.
     - Add a title (`Total Sales Revenue by Store in 2021`) and appropriate axis labels (`Store` and `Total Sales (USD)`).
     - Add data labels above each bar indicating the exact revenue.


6. **Line Plot:**
   - Create a line plot showing the sales revenue trends over time for a specific product (e.g., "Laptop").
   - Plot the `DATE` on the x-axis and `SALES` on the y-axis.
   - Customize the plot with the following:
     - Use **green** as the color of the line.
     - Add a title (`Sales Trends for Laptops in 2021`) and appropriate axis labels (`Date` and `Sales (USD)`).
     - Show a grid for better readability.
     - Include a legend indicating it represents "Laptop Sales".
