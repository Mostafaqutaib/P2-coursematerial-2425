## About the dataset

This dataset contains information about movies and TV shows available on Netflix. It provides a rich source of structured and unstructured data.

#### Features of the Dataset
* `show_id`: Unique identifier for the show or movie.
* `type`: Type of content
* `title`: Title of the movie or TV show.
* `director`: Name(s) of the director(s).
* `cast`: List of main cast members.
* `country`: Country or countries of origin where the content was produced.
* `date_added`: Date when the content was added to Netflix.
* `release_year`: Original release year of the content.
* `rating`: Age rating (e.g., TV-MA, PG, R) indicating suitability for audiences.
* `duration`: Duration of the content — minutes for movies, or number of seasons for shows.
* `listed_in`: List of genres or categories assigned by Netflix.
* `description`: Brief summary or promotional description of the content.
